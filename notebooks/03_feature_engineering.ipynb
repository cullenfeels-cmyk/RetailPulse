## Time-Based Feature Engineering

Time-based features are extracted from InvoiceDate to identify seasonal,
monthly, quarterly, and weekly purchasing patterns.

```python
# Create time-based features

df["Year"] = df["InvoiceDate"].dt.year
df["Month_Num"] = df["InvoiceDate"].dt.month
df["Month_Name"] = df["InvoiceDate"].dt.month_name()
df["Quarter"] = df["InvoiceDate"].dt.quarter
df["Day"] = df["InvoiceDate"].dt.day
df["Weekday"] = df["InvoiceDate"].dt.day_name()
df["Hour"] = df["InvoiceDate"].dt.hour

# Weekend indicator
df["Is_Weekend"] = df["InvoiceDate"].dt.dayofweek.isin([5, 6]).astype(int)

print("Time-based features created successfully ✅")
```

## RFM Feature Engineering

RFM (Recency, Frequency, Monetary) analysis helps identify valuable, loyal,
and at-risk customers based on their purchasing behaviour.

```python
# Create a reference date (1 day after the latest transaction)

snapshot_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)

print("Snapshot Date:", snapshot_date)
```

```python
rfm = df.groupby("Customer ID").agg({
    "InvoiceDate": lambda x: (snapshot_date - x.max()).days,
    "Invoice": "nunique",
    "Revenue": "sum"
})

rfm.columns = ["Recency", "Frequency", "Monetary"]

rfm.head()
```

```python
# Customer-level aggregations

customer_features = df.groupby("Customer ID").agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Invoice_Count=("Invoice", "nunique"),
    Unique_Products=("StockCode", "nunique")
)

customer_features.head()
```

```python
customer_features["Average_Order_Value"] = (
    customer_features["Total_Revenue"]
    / customer_features["Invoice_Count"]
)

customer_features.head()
```

```python
rfm = rfm.merge(
    customer_features,
    left_index=True,
    right_index=True
)

rfm.head()
```
